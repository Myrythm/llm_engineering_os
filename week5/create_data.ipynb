```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings

emmbedding = OpenAIEmbeddings(model="text-embedding-3-large")


# Load DB lama
db = Chroma(persist_directory="vector_db", embedding_function=emmbedding)

# Ambil seluruh isi (text + metadata)
data = db.get()

print(len(data["documents"]))  # jumlah chunk
print(data["documents"][0])    # isi teks chunk pertama
print(data["metadatas"][0])    # metadata-nya
```

```python
import json

# Simpan ke file JSONL
with open("chunks_from_db.jsonl", "w", encoding="utf-8") as f:
    for text, meta, id_ in zip(data["documents"], data["metadatas"], data["ids"]):
        record = {
            "id": id_,
            "text": text,
            "metadata": meta
        }
        f.write(json.dumps(record, ensure_ascii=False) + "\n")

print(f"Exported {len(data['documents'])} chunks to chunks_from_db.jsonl")
```
